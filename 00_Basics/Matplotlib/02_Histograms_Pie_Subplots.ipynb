# 📊 Matplotlib Tutorial 2: Histograms, Pie Charts & Subplots

**Objective:** Master distribution plots and multi-panel visualizations

In this notebook, you'll learn:
- Histograms for data distribution
- Pie charts for proportions
- Subplots for multiple visualizations
- Advanced customization techniques

---

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Load Titanic dataset
df = pd.read_csv('../titanic.csv')
df.head()
```

## 1. Histograms

Histograms show the distribution of continuous data by dividing it into bins

```python
# Basic histogram of passenger ages
fig, ax = plt.subplots(figsize=(10, 6))

ax.hist(df['Age'].dropna(), bins=30, color='skyblue', edgecolor='black', alpha=0.7)

ax.set_xlabel('Age', fontsize=14)
ax.set_ylabel('Frequency', fontsize=14)
ax.set_title('Distribution of Passenger Ages - Titanic', fontsize=16, fontweight='bold')
ax.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Multiple histograms with different bin sizes
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

ages = df['Age'].dropna()

# 10 bins
axes[0].hist(ages, bins=10, color='red', edgecolor='black', alpha=0.7)
axes[0].set_title('10 Bins', fontsize=14)
axes[0].set_xlabel('Age')
axes[0].set_ylabel('Count')
axes[0].grid(axis='y', alpha=0.3)

# 30 bins
axes[1].hist(ages, bins=30, color='green', edgecolor='black', alpha=0.7)
axes[1].set_title('30 Bins', fontsize=14)
axes[1].set_xlabel('Age')
axes[1].grid(axis='y', alpha=0.3)

# 50 bins
axes[2].hist(ages, bins=50, color='blue', edgecolor='black', alpha=0.7)
axes[2].set_title('50 Bins', fontsize=14)
axes[2].set_xlabel('Age')
axes[2].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Overlay histogram with KDE (Kernel Density Estimate)
fig, ax = plt.subplots(figsize=(10, 6))

# Histogram
counts, bins, patches = ax.hist(ages, bins=30, density=True, 
                                 color='skyblue', edgecolor='black', alpha=0.6, label='Histogram')

# KDE line
from scipy.stats import gaussian_kde
kde = gaussian_kde(ages)
x_kde = np.linspace(ages.min(), ages.max(), 200)
ax.plot(x_kde, kde(x_kde), color='red', linewidth=2, label='KDE')

ax.set_xlabel('Age', fontsize=14)
ax.set_ylabel('Density', fontsize=14)
ax.set_title('Age Distribution with KDE Overlay', fontsize=16, fontweight='bold')
ax.legend(fontsize=12)
ax.grid(axis='y', alpha=0.3)

plt.show()
```

## 2. Pie Charts

Pie charts show proportions of categories in a whole

```python
# Passenger class distribution
class_counts = df['Pclass'].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(10, 6))

colors = ['#FFD700', '#C0C0C0', '#CD7F32']
explode = (0.05, 0.05, 0.05)  # Slightly explode all slices

wedges, texts, autotexts = ax.pie(class_counts.values, 
                                    labels=['1st Class', '2nd Class', '3rd Class'],
                                    autopct='%1.1f%%',
                                    colors=colors,
                                    explode=explode,
                                    startangle=90,
                                    textprops={'fontsize': 12})

# Enhance text appearance
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')

ax.set_title('Passenger Distribution by Class', fontsize=16, fontweight='bold', pad=20)

plt.show()
```

```python
# Survival rate pie chart
survival_counts = df['Survived'].value_counts()

fig, ax = plt.subplots(figsize=(10, 6))

colors = ['#FF6B6B', '#4ECDC4']
explode = (0.05, 0)

wedges, texts, autotexts = ax.pie(survival_counts.values, 
                                    labels=['Not Survived', 'Survived'],
                                    autopct='%1.1f%%',
                                    colors=colors,
                                    explode=explode,
                                    startangle=90,
                                    textprops={'fontsize': 14})

for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')

ax.set_title('Titanic Survival Rate', fontsize=16, fontweight='bold', pad=20)

plt.show()
```

## 3. Subplots - Multiple Plots in One Figure

Subplots allow you to create complex multi-panel figures

```python
# 2x2 subplot grid
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Plot 1: Age histogram
axes[0, 0].hist(df['Age'].dropna(), bins=30, color='skyblue', edgecolor='black', alpha=0.7)
axes[0, 0].set_title('Age Distribution', fontsize=14, fontweight='bold')
axes[0, 0].set_xlabel('Age')
axes[0, 0].set_ylabel('Count')
axes[0, 0].grid(axis='y', alpha=0.3)

# Plot 2: Fare histogram
axes[0, 1].hist(df['Fare'], bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
axes[0, 1].set_title('Fare Distribution', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('Fare ($)')
axes[0, 1].set_ylabel('Count')
axes[0, 1].grid(axis='y', alpha=0.3)

# Plot 3: Survival by class
survival_by_class = df.groupby('Pclass')['Survived'].mean()
axes[1, 0].bar(survival_by_class.index, survival_by_class.values, 
               color=['#FFD700', '#C0C0C0', '#CD7F32'], edgecolor='black')
axes[1, 0].set_title('Survival Rate by Class', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('Passenger Class')
axes[1, 0].set_ylabel('Survival Rate')
axes[1, 0].set_xticks([1, 2, 3])
axes[1, 0].set_xticklabels(['1st', '2nd', '3rd'])
axes[1, 0].grid(axis='y', alpha=0.3)

# Plot 4: Gender distribution
gender_counts = df['Sex'].value_counts()
axes[1, 1].bar(gender_counts.index, gender_counts.values, 
               color=['#FF6B6B', '#4ECDC4'], edgecolor='black')
axes[1, 1].set_title('Gender Distribution', fontsize=14, fontweight='bold')
axes[1, 1].set_ylabel('Count')
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 4. Advanced Subplot Layouts

```python
# Using GridSpec for custom layouts
from matplotlib.gridspec import GridSpec

fig = plt.figure(figsize=(16, 10))
gs = GridSpec(2, 3, figure=fig)

# Large plot on left
ax1 = fig.add_subplot(gs[:, 0:2])
ax1.hist(df['Age'].dropna(), bins=30, color='skyblue', edgecolor='black', alpha=0.7)
ax1.set_title('Age Distribution (Large)', fontsize=14, fontweight='bold')
ax1.set_xlabel('Age')
ax1.set_ylabel('Count')
ax1.grid(axis='y', alpha=0.3)

# Small plots on right
ax2 = fig.add_subplot(gs[0, 2])
survival_counts = df['Survived'].value_counts()
ax2.pie(survival_counts.values, labels=['Died', 'Survived'], autopct='%1.1f%%',
        colors=['#FF6B6B', '#4ECDC4'], textprops={'fontsize': 10})
ax2.set_title('Survival', fontsize=12, fontweight='bold')

ax3 = fig.add_subplot(gs[1, 2])
gender_counts = df['Sex'].value_counts()
ax3.bar(gender_counts.index, gender_counts.values, color=['#FF6B6B', '#4ECDC4'])
ax3.set_title('Gender', fontsize=12, fontweight='bold')
ax3.set_ylabel('Count')

plt.tight_layout()
plt.show()
```

## 5. Annotations and Text

Add annotations to highlight important features

```python
fig, ax = plt.subplots(figsize=(10, 6))

# Histogram
counts, bins, patches = ax.hist(df['Age'].dropna(), bins=30, color='skyblue', 
                                 edgecolor='black', alpha=0.7)

# Find mean and median
mean_age = df['Age'].mean()
median_age = df['Age'].median()

# Add vertical lines
ax.axvline(mean_age, color='red', linewidth=2, linestyle='--', label=f'Mean: {mean_age:.1f}')
ax.axvline(median_age, color='green', linewidth=2, linestyle=':', label=f'Median: {median_age:.1f}')

# Add annotations
ax.annotate(f'Mean = {mean_age:.1f} years', 
            xy=(mean_age, max(counts)*0.9), 
            xytext=(mean_age + 5, max(counts)*0.8),
            fontsize=12, fontweight='bold',
            arrowprops=dict(arrowstyle='->', color='red'))

ax.set_xlabel('Age', fontsize=14)
ax.set_ylabel('Count', fontsize=14)
ax.set_title('Age Distribution with Statistics', fontsize=16, fontweight='bold')
ax.legend(fontsize=12)
ax.grid(axis='y', alpha=0.3)

plt.show()
```

## 6. Saving Figures

```python
# Create a plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(df['Age'].dropna(), bins=30, color='skyblue', edgecolor='black', alpha=0.7)
ax.set_title('Age Distribution', fontsize=16, fontweight='bold')
ax.set_xlabel('Age')
ax.set_ylabel('Count')

# Save with different formats and quality
plt.savefig('age_distribution.png', dpi=300, bbox_inches='tight')  # High quality PNG
plt.savefig('age_distribution.pdf', bbox_inches='tight')  # Vector format

print("✓ Figure saved as PNG (300 DPI)")
print("✓ Figure saved as PDF (vector)")

plt.show()
```
