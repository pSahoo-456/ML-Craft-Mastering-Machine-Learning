# 📊 Seaborn Tutorial 2: Categorical Plots

**Objective:** Master visualization of categorical data

In this notebook, you'll learn:
- Count plots for categorical frequencies
- Bar plots for aggregations
- Box plots and violin plots
- Point plots and swarm plots
- Factor plots for faceted categorical analysis

---

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline
sns.set_theme(style="whitegrid", font_scale=1.2)
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Load datasets
df = pd.read_csv('../titanic.csv')
tips = sns.load_dataset('tips')

print("Titanic:", df.shape)
print("Tips:", tips.shape)
```

## 1. Count Plots

Show the counts of observations in each categorical bin

```python
# Basic count plot
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Survival count
sns.countplot(data=df, x='Survived', ax=axes[0, 0])
axes[0, 0].set_title('Survival Count')
axes[0, 0].set_xticklabels(['Died', 'Survived'])

# Passenger class
sns.countplot(data=df, x='Pclass', ax=axes[0, 1])
axes[0, 1].set_title('Passenger Class Distribution')

# Gender distribution
sns.countplot(data=df, x='Sex', ax=axes[1, 0])
axes[1, 0].set_title('Gender Distribution')

# Embarkation port
sns.countplot(data=df, x='Embarked', ax=axes[1, 1])
axes[1, 1].set_title('Embarkation Port')

plt.tight_layout()
plt.show()
```

```python
# Count plot with hue
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Survival by class
sns.countplot(data=df, x='Pclass', hue='Survived', ax=axes[0])
axes[0].set_title('Survival Count by Class')
axes[0].set_xticklabels(['1st', '2nd', '3rd'])

# Survival by gender
sns.countplot(data=df, x='Sex', hue='Survived', ax=axes[1])
axes[1].set_title('Survival Count by Gender')

plt.tight_layout()
plt.show()
```

## 2. Bar Plots

Show point estimates and confidence intervals

```python
# Bar plot - survival rate by class
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Survival rate by class
sns.barplot(data=df, x='Pclass', y='Survived', ax=axes[0, 0])
axes[0, 0].set_title('Survival Rate by Class')
axes[0, 0].set_ylabel('Survival Rate')
axes[0, 0].set_xticklabels(['1st', '2nd', '3rd'])

# Survival rate by gender
sns.barplot(data=df, x='Sex', y='Survived', ax=axes[0, 1])
axes[0, 1].set_title('Survival Rate by Gender')
axes[0, 1].set_ylabel('Survival Rate')

# Average fare by class
sns.barplot(data=df, x='Pclass', y='Fare', ax=axes[1, 0])
axes[1, 0].set_title('Average Fare by Class')
axes[1, 0].set_ylabel('Average Fare ($)')
axes[1, 0].set_xticklabels(['1st', '2nd', '3rd'])

# Average age by class
sns.barplot(data=df, x='Pclass', y='Age', ax=axes[1, 1])
axes[1, 1].set_title('Average Age by Class')
axes[1, 1].set_ylabel('Average Age')
axes[1, 1].set_xticklabels(['1st', '2nd', '3rd'])

plt.tight_layout()
plt.show()
```

```python
# Bar plot with multiple hues
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Survival by class and gender
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', ax=axes[0])
axes[0].set_title('Survival Rate by Class and Gender')
axes[0].set_ylabel('Survival Rate')
axes[0].set_xticklabels(['1st', '2nd', '3rd'])

# Tips dataset example
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', ax=axes[1])
axes[1].set_title('Average Bill by Day and Gender')
axes[1].set_ylabel('Average Total Bill ($)')

plt.tight_layout()
plt.show()
```

## 3. Box Plots

Show distribution based on quartiles and outliers

```python
# Box plots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Fare by class
sns.boxplot(data=df, x='Pclass', y='Fare', ax=axes[0, 0])
axes[0, 0].set_title('Fare by Class')
axes[0, 0].set_xticklabels(['1st', '2nd', '3rd'])

# Age by class
sns.boxplot(data=df, x='Pclass', y='Age', ax=axes[0, 1])
axes[0, 1].set_title('Age by Class')
axes[0, 1].set_xticklabels(['1st', '2nd', '3rd'])

# Age by survival
sns.boxplot(data=df, x='Survived', y='Age', ax=axes[1, 0])
axes[1, 0].set_title('Age by Survival')
axes[1, 0].set_xticklabels(['Died', 'Survived'])

# Tips example
sns.boxplot(data=tips, x='day', y='total_bill', hue='sex', ax=axes[1, 1])
axes[1, 1].set_title('Total Bill by Day and Gender')

plt.tight_layout()
plt.show()
```

## 4. Violin Plots

Combine box plot with KDE to show distribution shape

```python
# Violin plots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Fare by class
sns.violinplot(data=df, x='Pclass', y='Fare', ax=axes[0, 0])
axes[0, 0].set_title('Fare Distribution by Class')
axes[0, 0].set_xticklabels(['1st', '2nd', '3rd'])

# Age by class with split violins
sns.violinplot(data=df, x='Pclass', y='Age', hue='Survived', split=True, ax=axes[0, 1])
axes[0, 1].set_title('Age by Class and Survival')
axes[0, 1].set_xticklabels(['1st', '2nd', '3rd'])

# Inner quartile
sns.violinplot(data=df, x='Pclass', y='Age', inner='quartile', ax=axes[1, 0])
axes[1, 0].set_title('Age by Class (Quartiles)')
axes[1, 0].set_xticklabels(['1st', '2nd', '3rd'])

# Tips example
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex', split=True, ax=axes[1, 1])
axes[1, 1].set_title('Total Bill by Day and Gender')

plt.tight_layout()
plt.show()
```

## 5. Point Plots

Show point estimates with confidence intervals using lines

```python
# Point plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Survival rate trend by class
sns.pointplot(data=df, x='Pclass', y='Survived', ax=axes[0])
axes[0].set_title('Survival Rate by Class')
axes[0].set_ylabel('Survival Rate')
axes[0].set_xticklabels(['1st', '2nd', '3rd'])

# With hue
sns.pointplot(data=df, x='Pclass', y='Survived', hue='Sex', ax=axes[1])
axes[1].set_title('Survival Rate by Class and Gender')
axes[1].set_ylabel('Survival Rate')
axes[1].set_xticklabels(['1st', '2nd', '3rd'])

plt.tight_layout()
plt.show()
```

## 6. Strip and Swarm Plots

Show individual data points

```python
# Strip and swarm plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Strip plot (jittered scatter)
sns.stripplot(data=tips, x='day', y='total_bill', ax=axes[0], alpha=0.5)
axes[0].set_title('Strip Plot - Total Bill by Day')

# Swarm plot (non-overlapping)
sns.swarmplot(data=tips, x='day', y='total_bill', ax=axes[1], alpha=0.6)
axes[1].set_title('Swarm Plot - Total Bill by Day')

plt.tight_layout()
plt.show()
```

```python
# Combine box plot with swarm plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Box + strip
sns.boxplot(data=tips, x='day', y='total_bill', ax=axes[0])
sns.stripplot(data=tips, x='day', y='total_bill', color='black', 
             alpha=0.3, ax=axes[0])
axes[0].set_title('Box Plot + Strip Plot')

# Violin + swarm
sns.violinplot(data=tips, x='day', y='total_bill', inner=None, ax=axes[1])
sns.swarmplot(data=tips, x='day', y='total_bill', color='black', 
             alpha=0.5, ax=axes[1])
axes[1].set_title('Violin Plot + Swarm Plot')

plt.tight_layout()
plt.show()
```

## 7. Factor Plot (catplot)

Figure-level function for categorical plots with faceting

```python
# Catplot with count plots
g = sns.catplot(data=df, x='Pclass', hue='Survived', col='Sex',
                kind='count', height=5, aspect=1)
g.fig.suptitle('Survival Count by Class and Gender', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# Catplot with box plots
g = sns.catplot(data=df, x='Pclass', y='Age', hue='Survived',
                kind='box', height=5, aspect=1)
g.fig.suptitle('Age by Class and Survival', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# Catplot with violin plots and col/row faceting
g = sns.catplot(data=tips, x='day', y='total_bill', hue='sex',
                col='time', row='smoker', kind='violin', split=True,
                height=4, aspect=1)
g.fig.suptitle('Total Bill Distribution', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

## 8. Comprehensive Categorical Analysis - Titanic

```python
# Complete categorical analysis dashboard
fig = plt.figure(figsize=(20, 12))
fig.suptitle('TITANIC - CATEGORICAL ANALYSIS', fontsize=24, fontweight='bold', y=0.98)

gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

# Row 1: Count plots
ax1 = fig.add_subplot(gs[0, 0])
sns.countplot(data=df, x='Survived', ax=ax1)
ax1.set_title('Survival Count', fontsize=14)
ax1.set_xticklabels(['Died', 'Survived'])

ax2 = fig.add_subplot(gs[0, 1])
sns.countplot(data=df, x='Pclass', ax=ax2)
ax2.set_title('Class Distribution', fontsize=14)
ax2.set_xticklabels(['1st', '2nd', '3rd'])

ax3 = fig.add_subplot(gs[0, 2])
sns.countplot(data=df, x='Sex', ax=ax3)
ax3.set_title('Gender Distribution', fontsize=14)

# Row 2: Bar plots
ax4 = fig.add_subplot(gs[1, 0])
sns.barplot(data=df, x='Pclass', y='Survived', ax=ax4)
ax4.set_title('Survival Rate by Class', fontsize=14)
ax4.set_ylabel('Survival Rate')
ax4.set_xticklabels(['1st', '2nd', '3rd'])

ax5 = fig.add_subplot(gs[1, 1])
sns.barplot(data=df, x='Sex', y='Survived', ax=ax5)
ax5.set_title('Survival Rate by Gender', fontsize=14)
ax5.set_ylabel('Survival Rate')

ax6 = fig.add_subplot(gs[1, 2])
sns.barplot(data=df, x='Pclass', y='Fare', ax=ax6)
ax6.set_title('Average Fare by Class', fontsize=14)
ax6.set_ylabel('Average Fare ($)')
ax6.set_xticklabels(['1st', '2nd', '3rd'])

# Row 3: Box and violin plots
ax7 = fig.add_subplot(gs[2, 0])
sns.boxplot(data=df, x='Pclass', y='Age', ax=ax7)
ax7.set_title('Age by Class', fontsize=14)
ax7.set_xticklabels(['1st', '2nd', '3rd'])

ax8 = fig.add_subplot(gs[2, 1])
sns.violinplot(data=df, x='Pclass', y='Fare', ax=ax8)
ax8.set_title('Fare by Class', fontsize=14)
ax8.set_xticklabels(['1st', '2nd', '3rd'])

ax9 = fig.add_subplot(gs[2, 2])
sns.pointplot(data=df, x='Pclass', y='Survived', hue='Sex', ax=ax9)
ax9.set_title('Survival by Class & Gender', fontsize=14)
ax9.set_ylabel('Survival Rate')
ax9.set_xticklabels(['1st', '2nd', '3rd'])

plt.tight_layout()
plt.savefig('titanic_categorical_analysis.png', dpi=300, bbox_inches='tight')
print("✓ Saved as 'titanic_categorical_analysis.png'")
plt.show()
```

## 9. Key Takeaways

✅ **countplot**: Show category frequencies  
✅ **barplot**: Show aggregations with confidence intervals  
✅ **boxplot**: Show quartiles and outliers  
✅ **violinplot**: Show distribution shape + quartiles  
✅ **pointplot**: Show trends with confidence intervals  
✅ **stripplot/swarmplot**: Show individual observations  
✅ **catplot**: Figure-level function for faceted categorical plots
