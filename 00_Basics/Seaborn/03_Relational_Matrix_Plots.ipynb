```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline
sns.set_theme(style="whitegrid", font_scale=1.2)
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Load datasets
df = pd.read_csv('../titanic.csv')
tips = sns.load_dataset('tips')
flights = sns.load_dataset('flights')

print("Titanic:", df.shape)
print("Tips:", tips.shape)
print("Flights:", flights.shape)
```

## 1. Scatter Plots with scatterplot

Show relationships between two numerical variables

```python
# Basic scatter plots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Simple scatter
sns.scatterplot(data=tips, x='total_bill', y='tip', ax=axes[0, 0])
axes[0, 0].set_title('Tip vs Total Bill')

# With hue
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', ax=axes[0, 1])
axes[0, 1].set_title('Tip vs Bill by Time')

# With size
sns.scatterplot(data=tips, x='total_bill', y='tip', size='size', ax=axes[1, 0])
axes[1, 0].set_title('Tip vs Bill by Party Size')

# With hue and style
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='day', style='time', ax=axes[1, 1])
axes[1, 1].set_title('Tip vs Bill by Day and Time')

plt.tight_layout()
plt.show()
```

```python
# Titanic scatter plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Age vs Fare
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived', 
               palette=['red', 'green'], alpha=0.6, ax=axes[0])
axes[0].set_title('Age vs Fare by Survival')

# With different markers
sns.scatterplot(data=df, x='Age', y='Fare', hue='Sex', 
               style='Survived', alpha=0.6, ax=axes[1])
axes[1].set_title('Age vs Fare by Gender and Survival')

plt.tight_layout()
plt.show()
```

## 2. Line Plots with lineplot

Show trends and relationships over continuous variables

```python
# Line plots with flights dataset
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Simple line plot
sns.lineplot(data=flights, x='year', y='passengers', ax=axes[0, 0])
axes[0, 0].set_title('Passengers Over Time')

# With confidence interval
sns.lineplot(data=flights, x='year', y='passengers', ci=95, ax=axes[0, 1])
axes[0, 1].set_title('Passengers with 95% CI')

# Multiple lines by month
flights_pivot = flights.pivot_table(index='year', columns='month', values='passengers')
for month in flights_pivot.columns[:4]:  # First 4 months
    sns.lineplot(x=flights_pivot.index, y=flights_pivot[month], label=month, ax=axes[1, 0])
axes[1, 0].set_title('Passengers by Month (First 4)')
axes[1, 0].legend()

# Tips dataset - trend
tips_avg = tips.groupby('size')['total_bill'].mean().reset_index()
sns.lineplot(data=tips_avg, x='size', y='total_bill', marker='o', ax=axes[1, 1])
axes[1, 1].set_title('Average Bill by Party Size')

plt.tight_layout()
plt.show()
```

## 3. Relplot - Figure-Level Relational Plot

Create faceted relational plots easily

```python
# Relplot with scatter
g = sns.relplot(data=tips, x='total_bill', y='tip', hue='day', 
               col='time', kind='scatter', height=5, aspect=1)
g.fig.suptitle('Tip vs Bill by Time', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# Relplot with line
g = sns.relplot(data=flights, x='year', y='passengers', kind='line',
               col='month', col_wrap=4, height=3, aspect=1)
g.fig.suptitle('Passenger Trends by Month', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# Relplot with size and hue
g = sns.relplot(data=tips, x='total_bill', y='tip', hue='size', 
               size='size', kind='scatter', height=6, aspect=1.2)
g.fig.suptitle('Tip vs Bill with Party Size', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

## 4. Heatmaps

Visualize matrix data with colors

```python
# Correlation heatmap
numeric_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
corr_matrix = df[numeric_cols].corr()

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Basic heatmap
sns.heatmap(corr_matrix, annot=True, cmap='RdBu_r', center=0, ax=axes[0])
axes[0].set_title('Correlation Matrix', fontsize=14)

# With different colormap
sns.heatmap(corr_matrix, annot=True, cmap='viridis', ax=axes[1])
axes[1].set_title('Correlation Matrix (viridis)', fontsize=14)

plt.tight_layout()
plt.show()
```

```python
# Flights heatmap
flights_pivot = flights.pivot_table(index='month', columns='year', values='passengers')

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Basic heatmap
sns.heatmap(flights_pivot, annot=True, fmt='d', cmap='YlGnBu', ax=axes[0])
axes[0].set_title('Passengers by Month and Year', fontsize=14)

# With different options
sns.heatmap(flights_pivot, annot=False, cmap='YlOrRd', linewidths=0.5, ax=axes[1])
axes[1].set_title('Passengers Heatmap (no annotations)', fontsize=14)

plt.tight_layout()
plt.show()
```

## 5. Clustermap

Hierarchically clustered heatmap

```python
# Clustermap of flights data
g = sns.clustermap(flights_pivot, cmap='YlGnBu', figsize=(12, 8))
g.fig.suptitle('Clustermap - Passengers by Month and Year', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# Clustermap with standard scaling
from scipy.cluster.hierarchy import linkage

# Standardize the data
flights_scaled = (flights_pivot - flights_pivot.mean()) / flights_pivot.std()

g = sns.clustermap(flights_scaled, cmap='RdBu_r', center=0, figsize=(12, 8))
g.fig.suptitle('Clustermap - Standardized Passenger Data', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

## 6. Pairplot - Comprehensive Pairwise Relationships

Visualize all pairwise relationships in a dataset

```python
# Basic pairplot
penguins = sns.load_dataset('penguins')
penguins_clean = penguins.dropna()

g = sns.pairplot(penguins_clean, hue='species', diag_kind='hist',
                plot_kws={'alpha': 0.6}, height=2.5)
g.fig.suptitle('Penguins Dataset - Pair Plot', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# Pairplot with different plot types
g = sns.pairplot(penguins_clean, hue='species', diag_kind='kde',
                kind='reg', plot_kws={'scatter_kws': {'alpha': 0.5}}, height=2.5)
g.fig.suptitle('Penguins - Pair Plot with Regression', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

## 7. LM Plot - Linear Model Plot

Scatter plot with regression line

```python
# LM plot
g = sns.lmplot(data=tips, x='total_bill', y='tip', hue='time',
              height=6, aspect=1.2)
g.fig.suptitle('Tip vs Bill with Regression Lines', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# LM plot with facets
g = sns.lmplot(data=tips, x='total_bill', y='tip', col='day',
              hue='smoker', height=5, aspect=1)
g.fig.suptitle('Tip vs Bill by Day and Smoking Status', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

## 8. Complete EDA Dashboard - Titanic Dataset

```python
# Comprehensive EDA with Seaborn
fig = plt.figure(figsize=(20, 15))
fig.suptitle('TITANIC DATASET - COMPLETE EDA WITH SEABORN', 
             fontsize=24, fontweight='bold', y=0.98)

gs = fig.add_gridspec(4, 3, hspace=0.3, wspace=0.3)

# Row 1: Distributions
ax1 = fig.add_subplot(gs[0, 0])
sns.histplot(data=df, x='Age', kde=True, ax=ax1)
ax1.set_title('Age Distribution', fontsize=14)

ax2 = fig.add_subplot(gs[0, 1])
sns.histplot(data=df, x='Fare', kde=True, ax=ax2)
ax2.set_title('Fare Distribution', fontsize=14)

ax3 = fig.add_subplot(gs[0, 2])
sns.countplot(data=df, x='Survived', ax=ax3)
ax3.set_title('Survival Count', fontsize=14)
ax3.set_xticklabels(['Died', 'Survived'])

# Row 2: Categorical analysis
ax4 = fig.add_subplot(gs[1, 0])
sns.barplot(data=df, x='Pclass', y='Survived', ax=ax4)
ax4.set_title('Survival Rate by Class', fontsize=14)
ax4.set_ylabel('Survival Rate')
ax4.set_xticklabels(['1st', '2nd', '3rd'])

ax5 = fig.add_subplot(gs[1, 1])
sns.barplot(data=df, x='Sex', y='Survived', ax=ax5)
ax5.set_title('Survival Rate by Gender', fontsize=14)
ax5.set_ylabel('Survival Rate')

ax6 = fig.add_subplot(gs[1, 2])
sns.countplot(data=df, x='Pclass', hue='Survived', ax=ax6)
ax6.set_title('Survival by Class', fontsize=14)
ax6.set_xticklabels(['1st', '2nd', '3rd'])

# Row 3: Box and violin plots
ax7 = fig.add_subplot(gs[2, 0])
sns.boxplot(data=df, x='Pclass', y='Age', ax=ax7)
ax7.set_title('Age by Class', fontsize=14)
ax7.set_xticklabels(['1st', '2nd', '3rd'])

ax8 = fig.add_subplot(gs[2, 1])
sns.violinplot(data=df, x='Pclass', y='Fare', ax=ax8)
ax8.set_title('Fare by Class', fontsize=14)
ax8.set_xticklabels(['1st', '2nd', '3rd'])

ax9 = fig.add_subplot(gs[2, 2])
sns.pointplot(data=df, x='Pclass', y='Survived', hue='Sex', ax=ax9)
ax9.set_title('Survival by Class & Gender', fontsize=14)
ax9.set_ylabel('Survival Rate')
ax9.set_xticklabels(['1st', '2nd', '3rd'])

# Row 4: Correlation and scatter
ax10 = fig.add_subplot(gs[3, 0])
numeric_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
corr_matrix = df[numeric_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='RdBu_r', center=0, ax=ax10)
ax10.set_title('Correlation Matrix', fontsize=14)

ax11 = fig.add_subplot(gs[3, 1])
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived', 
               palette=['red', 'green'], alpha=0.5, ax=ax11)
ax11.set_title('Age vs Fare by Survival', fontsize=14)

ax12 = fig.add_subplot(gs[3, 2])
sns.kdeplot(data=df, x='Age', hue='Survived', fill=True, 
           palette=['red', 'green'], alpha=0.5, ax=ax12)
ax12.set_title('Age Density by Survival', fontsize=14)

plt.tight_layout()
plt.savefig('titanic_complete_eda_seaborn.png', dpi=300, bbox_inches='tight')
print("✓ Saved as 'titanic_complete_eda_seaborn.png'")
plt.show()
```

## 9. Key Insights from Titanic EDA

### 📊 Major Findings:

1. **Gender was strongest predictor** - Females had ~74% survival vs males ~19%
2. **Class mattered** - 1st class: 63%, 2nd: 47%, 3rd: 24% survival
3. **Children prioritized** - Young passengers had higher survival
4. **Fare correlates with survival** - Higher fare = better survival (linked to class)
5. **Family size impact** - Small families (2-4) had better survival
