```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline

# Set default style
sns.set_theme(style="whitegrid", font_scale=1.2)
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Why Seaborn?

Seaborn provides:
- **Simpler syntax** for complex plots
- **Built-in themes** and color palettes
- **Statistical aggregations** automatically
- **DataFrame integration** - works directly with pandas
- **Beautiful defaults** out of the box

```python
# Load datasets
df = pd.read_csv('../titanic.csv')

# Also load built-in seaborn datasets for examples
tips = sns.load_dataset('tips')
penguins = sns.load_dataset('penguins')

print("Titanic dataset:", df.shape)
print("Tips dataset:", tips.shape)
print("Penguins dataset:", penguins.shape)
```

## 2. Seaborn vs Matplotlib - Quick Comparison

```python
# Matplotlib approach (verbose)
fig, ax = plt.subplots(figsize=(8, 5))

for species in penguins['species'].unique():
    subset = penguins[penguins['species'] == species]
    ax.scatter(subset['bill_length_mm'], subset['bill_depth_mm'], 
              label=species, alpha=0.6)

ax.set_xlabel('Bill Length (mm)')
ax.set_ylabel('Bill Depth (mm)')
ax.set_title('Matplotlib Approach')
ax.legend()
plt.show()
```

```python
# Seaborn approach (concise)
sns.scatterplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', 
                hue='species', alpha=0.6)
plt.title('Seaborn Approach - Much Simpler!')
plt.show()
```

## 3. Histograms with histplot

Modern replacement for plt.hist() with more features

```python
# Basic histogram
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Simple histogram
sns.histplot(data=df, x='Age', bins=30, ax=axes[0, 0])
axes[0, 0].set_title('Basic Histogram')

# With KDE overlay
sns.histplot(data=df, x='Age', bins=30, kde=True, ax=axes[0, 1])
axes[0, 1].set_title('Histogram with KDE')

# Colored by survival
sns.histplot(data=df, x='Age', bins=30, hue='Survived', multiple='stack', ax=axes[1, 0])
axes[1, 0].set_title('Stacked by Survival')

# Overlapping with transparency
sns.histplot(data=df, x='Age', bins=30, hue='Survived', multiple='layer', 
             palette=['red', 'green'], alpha=0.6, ax=axes[1, 1])
axes[1, 1].set_title('Overlapping Histograms')

plt.tight_layout()
plt.show()
```

```python
# Distribution with statistics
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Fare distribution
sns.histplot(data=df, x='Fare', bins=50, kde=True, color='skyblue', ax=axes[0])
axes[0].axvline(df['Fare'].mean(), color='red', linestyle='--', 
               label=f'Mean: ${df["Fare"].mean():.2f}')
axes[0].axvline(df['Fare'].median(), color='green', linestyle=':',
               label=f'Median: ${df["Fare"].median():.2f}')
axes[0].set_title('Fare Distribution')
axes[0].legend()

# Age distribution by class
sns.histplot(data=df, x='Age', hue='Pclass', multiple='stack', 
             palette=['gold', 'silver', '#CD7F32'], ax=axes[1])
axes[1].set_title('Age Distribution by Class')

plt.tight_layout()
plt.show()
```

## 4. Kernel Density Estimation (KDE)

Smooth estimate of the probability density function

```python
# KDE plots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Single KDE
sns.kdeplot(data=df, x='Age', fill=True, ax=axes[0, 0])
axes[0, 0].set_title('Age KDE')

# Multiple KDEs
sns.kdeplot(data=df, x='Age', hue='Survived', fill=True, 
            palette=['red', 'green'], alpha=0.5, ax=axes[0, 1])
axes[0, 1].set_title('Age KDE by Survival')

# Vertical KDE
sns.kdeplot(data=df, y='Fare', fill=True, ax=axes[1, 0])
axes[1, 0].set_title('Fare KDE (Vertical)')

# Multiple vertical KDEs
for pclass in [1, 2, 3]:
    subset = df[df['Pclass'] == pclass]
    sns.kdeplot(data=subset, y='Fare', label=f'Class {pclass}', ax=axes[1, 1])
axes[1, 1].set_title('Fare KDE by Class')
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```

## 5. Distribution Plot (displot) - Figure-Level Function

displot is a figure-level function that can create multiple types of distribution plots

```python
# displot with different kinds
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Histogram
sns.histplot(data=df, x='Age', kde=True, ax=axes[0])
axes[0].set_title('histplot')

# KDE only
sns.kdeplot(data=df, x='Age', fill=True, ax=axes[1])
axes[1].set_title('kdeplot')

# ECDF (Empirical Cumulative Distribution Function)
sns.ecdfplot(data=df, x='Age', ax=axes[2])
axes[2].set_title('ecdfplot')

plt.tight_layout()
plt.show()
```

```python
# displot - figure-level function with facets
g = sns.displot(data=df, x='Age', hue='Survived', col='Sex', 
                kind='hist', kde=True, height=5, aspect=1.2)
g.fig.suptitle('Age Distribution by Survival and Gender', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

## 6. Rug Plot

Shows individual observations as ticks along the axis

```python
# Histogram with rug plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# With rug
sns.histplot(data=df, x='Age', kde=True, ax=axes[0])
sns.rugplot(data=df, x='Age', ax=axes[0])
axes[0].set_title('Histogram with Rug Plot')

# KDE with rug
sns.kdeplot(data=df, x='Age', fill=True, ax=axes[1])
sns.rugplot(data=df, x='Age', ax=axes[1], height=0.05)
axes[1].set_title('KDE with Rug Plot')

plt.tight_layout()
plt.show()
```

## 7. Joint Plot

Bivariate distribution with marginal distributions

```python
# Joint plot - scatter with marginal histograms
g = sns.jointplot(data=df, x='Age', y='Fare', kind='scatter', 
                  hue='Survived', palette=['red', 'green'], alpha=0.6,
                  height=8)
g.fig.suptitle('Age vs Fare with Survival', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# Joint plot with regression
g = sns.jointplot(data=df, x='Age', y='Fare', kind='reg',
                  scatter_kws={'alpha': 0.3}, height=8)
g.fig.suptitle('Age vs Fare with Regression Line', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# Joint plot with KDE
g = sns.jointplot(data=df, x='Age', y='Fare', kind='kde',
                  fill=True, cmap='Blues', height=8)
g.fig.suptitle('Age vs Fare - Joint KDE', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

## 8. Pair Plot

Multiple pairwise relationships in a dataset

```python
# Pair plot with penguins dataset
penguins_clean = penguins.dropna()
g = sns.pairplot(penguins_clean, hue='species', diag_kind='hist',
                 plot_kws={'alpha': 0.6}, height=2.5)
g.fig.suptitle('Penguins Dataset - Pair Plot', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

## 9. Styling and Themes

```python
# Different seaborn themes
themes = ['white', 'whitegrid', 'dark', 'darkgrid', 'ticks']

fig, axes = plt.subplots(1, 5, figsize=(20, 4))

for ax, theme in zip(axes, themes):
    with sns.axes_style(theme):
        sns.histplot(data=df, x='Age', kde=True, ax=ax)
        ax.set_title(f'{theme} theme', fontsize=12)
        ax.set_xlabel('')
        ax.set_ylabel('')

plt.tight_layout()
plt.show()
```

```python
# Color palettes
palettes = ['deep', 'muted', 'pastel', 'bright', 'dark', 'colorblind']

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for ax, palette in zip(axes.flat, palettes):
    sns.histplot(data=df, x='Age', hue='Pclass', multiple='stack',
                 palette=palette, ax=ax)
    ax.set_title(f'{palette} palette', fontsize=14)
    ax.set_xlabel('')

plt.tight_layout()
plt.show()
```

## 10. Key Takeaways

✅ **Seaborn simplifies complex plots** with DataFrame integration  
✅ **histplot** is the modern histogram function  
✅ **kdeplot** shows smooth density estimates  
✅ **displot** is a figure-level function for distributions  
✅ **jointplot** shows bivariate + marginal distributions  
✅ **pairplot** visualizes all pairwise relationships  
✅ **Themes and palettes** make beautiful plots easily
